# Phase 2: Data Understanding and EDA



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import zipfile
import time
from sklearn.model_selection import train_test_split, GridSearchCV, learning_curve
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, LogisticRegression, Ridge, Lasso
from sklearn.tree import DecisionTreeRegressor, DecisionTreeClassifier, plot_tree, export_text
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix, classification_report
from sklearn.inspection import permutation_importance
with zipfile.ZipFile("archive.zip") as z:
    with z.open("Housing.csv") as f:
        housing = pd.read_csv(f)
with zipfile.ZipFile("archive (1).zip") as z:
    with z.open("creditcard.csv") as f:
        credit = pd.read_csv(f)
print("Housing Dataset Shape:")
print(housing.shape)
print("\nCredit Card Dataset Shape:")
print(credit.shape)
print("\nHousing First 5 Rows:")
display(housing.head())
print("\nCredit Card First 5 Rows:")
display(credit.head())
print("\nHousing Info:")
print(housing.info())
print("\nCredit Card Info:")
print(credit.info())
print("\nHousing Missing Values:")
print(housing.isnull().sum())
print("\nCredit Card Missing Values:")
print(credit.isnull().sum())
print("\nHousing Duplicate Rows:")
print(housing.duplicated().sum())
print("\nCredit Card Duplicate Rows:")
print(credit.duplicated().sum())
print("\nHousing Numerical Summary:")
display(housing.describe())
print("\nCredit Card Numerical Summary:")
display(credit.describe())
housing_num_cols = housing.select_dtypes(include=["int64", "float64"]).columns
for col in housing_num_cols:
    plt.figure(figsize=(6, 4))
    housing[col].hist(bins=30)
    plt.title("Housing Distribution of " + col)
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.show()
housing_cat_cols = housing.select_dtypes(include=["object"]).columns
for col in housing_cat_cols:
    plt.figure(figsize=(6, 4))
    housing[col].value_counts().plot(kind="bar")
    plt.title("Housing Count Plot of " + col)
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.xticks(rotation=45)
    plt.show()
plt.figure(figsize=(8, 6))
housing_corr = housing[housing_num_cols].corr()
plt.imshow(housing_corr)
plt.colorbar()
plt.xticks(range(len(housing_corr.columns)), housing_corr.columns, rotation=90)
plt.yticks(range(len(housing_corr.columns)), housing_corr.columns)
plt.title("Housing Correlation Heatmap")
plt.show()
plt.figure(figsize=(6, 4))
plt.scatter(housing["area"], housing["price"])
plt.title("Area vs Price")
plt.xlabel("Area")
plt.ylabel("Price")
plt.show()
print("\nHousing Outlier Detection using IQR:")
for col in housing_num_cols:
    q1 = housing[col].quantile(0.25)
    q3 = housing[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outliers = housing[(housing[col] < lower) | (housing[col] > upper)].shape[0]
    print(col, ":", outliers)
print("\nCredit Card Class Distribution:")
print(credit["Class"].value_counts())
print(credit["Class"].value_counts(normalize=True) * 100)
plt.figure(figsize=(5, 4))
credit["Class"].value_counts().plot(kind="bar")
plt.title("Credit Card Fraud Class Distribution")
plt.xlabel("Class")
plt.ylabel("Count")
plt.show()
plt.figure(figsize=(6, 4))
credit["Amount"].hist(bins=50)
plt.title("Transaction Amount Distribution")
plt.xlabel("Amount")
plt.ylabel("Count")
plt.show()
plt.figure(figsize=(8, 6))
credit_corr = credit[["Time", "Amount", "Class"]].corr()
plt.imshow(credit_corr)
plt.colorbar()
plt.xticks(range(len(credit_corr.columns)), credit_corr.columns, rotation=90)
plt.yticks(range(len(credit_corr.columns)), credit_corr.columns)
plt.title("Credit Card Correlation Heatmap")
plt.show()
print("\nEDA Interpretation:")
print("Housing price is affected by numerical and categorical house features.")
print("Credit card fraud data is highly imbalanced because fraud cases are much fewer than normal transactions.")
```

# Phase 3: Data Preprocessing

```python
housing_clean = housing.copy()
credit_clean = credit.copy()
housing_clean = housing_clean.drop_duplicates()
credit_clean = credit_clean.drop_duplicates()
for col in housing_clean.select_dtypes(include=["int64", "float64"]).columns:
    housing_clean[col] = housing_clean[col].fillna(housing_clean[col].median())
for col in housing_clean.select_dtypes(include=["object"]).columns:
    housing_clean[col] = housing_clean[col].fillna(housing_clean[col].mode()[0])
for col in credit_clean.columns:
    credit_clean[col] = credit_clean[col].fillna(credit_clean[col].median())
housing_clean["total_rooms"] = housing_clean["bedrooms"] + housing_clean["bathrooms"]
housing_clean["area_per_bedroom"] = housing_clean["area"] / housing_clean["bedrooms"]
housing_clean["luxury_score"] = (
    (housing_clean["airconditioning"] == "yes").astype(int)
    + (housing_clean["hotwaterheating"] == "yes").astype(int)
    + (housing_clean["prefarea"] == "yes").astype(int)
)
for col in ["price", "area"]:
    q1 = housing_clean[col].quantile(0.25)
    q3 = housing_clean[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    housing_clean[col] = housing_clean[col].clip(lower, upper)
X_reg = housing_clean.drop("price", axis=1)
y_reg = housing_clean["price"]
X_reg = pd.get_dummies(X_reg, drop_first=True)
X_reg_train, X_reg_test, y_reg_train, y_reg_test = train_test_split(
    X_reg, y_reg, test_size=0.20, random_state=42
)
reg_scaler = StandardScaler()
X_reg_train_scaled = reg_scaler.fit_transform(X_reg_train)
X_reg_test_scaled = reg_scaler.transform(X_reg_test)
fraud = credit_clean[credit_clean["Class"] == 1]
normal = credit_clean[credit_clean["Class"] == 0].sample(n=min(10000, (credit_clean["Class"] == 0).sum()), random_state=42)
credit_sample = pd.concat([fraud, normal])
credit_sample = credit_sample.sample(frac=1, random_state=42)
X_cls = credit_sample.drop("Class", axis=1)
y_cls = credit_sample["Class"]
X_cls_train, X_cls_test, y_cls_train, y_cls_test = train_test_split(
    X_cls, y_cls, test_size=0.20, random_state=42, stratify=y_cls
)
cls_scaler = StandardScaler()
X_cls_train_scaled = cls_scaler.fit_transform(X_cls_train)
X_cls_test_scaled = cls_scaler.transform(X_cls_test)
print("Regression Training Shape:")
print(X_reg_train.shape)
print("\nRegression Testing Shape:")
print(X_reg_test.shape)
print("\nClassification Training Shape:")
print(X_cls_train.shape)
print("\nClassification Testing Shape:")
print(X_cls_test.shape)
print("\nClassification Class Balance:")
print(y_cls.value_counts(normalize=True) * 100)
print("\nPreprocessing Explanation:")
print("Missing values were filled using median for numerical columns and mode for categorical columns.")
print("Categorical variables were encoded using get_dummies.")
print("Scaling was applied for Linear Regression, Ridge, Lasso, and Logistic Regression.")
print("Outliers in housing price and area were capped using IQR.")
print("Classification split used stratification to preserve class ratio.")
```

# Phase 4: Regression Modeling


```python
regression_results = []
def evaluate_regression(model_name, model, X_train_data, X_test_data):
    start_time = time.time()
    model.fit(X_train_data, y_reg_train)
    training_time = time.time() - start_time
    y_pred = model.predict(X_test_data)
    mae = mean_absolute_error(y_reg_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_reg_test, y_pred))
    r2 = r2_score(y_reg_test, y_pred)
    regression_results.append([model_name, mae, rmse, r2, training_time])
    print("\n", model_name)
    print("MAE:", mae)
    print("RMSE:", rmse)
    print("R2 Score:", r2)
    print("Training Time:", training_time)
    return model
linear_model = evaluate_regression(
    "Linear Regression",
    LinearRegression(),
    X_reg_train_scaled,
    X_reg_test_scaled
)
tree_reg = DecisionTreeRegressor(random_state=42)
tree_params = {
    "max_depth": [3, 5, 7, 10],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 5]
}
tree_reg_grid = GridSearchCV(tree_reg, tree_params, cv=3, scoring="r2")
tree_reg_grid.fit(X_reg_train, y_reg_train)
best_tree_reg = tree_reg_grid.best_estimator_
tree_reg_pred = best_tree_reg.predict(X_reg_test)
tree_reg_mae = mean_absolute_error(y_reg_test, tree_reg_pred)
tree_reg_rmse = np.sqrt(mean_squared_error(y_reg_test, tree_reg_pred))
tree_reg_r2 = r2_score(y_reg_test, tree_reg_pred)
regression_results.append(["Decision Tree Regressor", tree_reg_mae, tree_reg_rmse, tree_reg_r2, 0])
print("\nDecision Tree Regressor")
print("Best Parameters:", tree_reg_grid.best_params_)
print("MAE:", tree_reg_mae)
print("RMSE:", tree_reg_rmse)
print("R2 Score:", tree_reg_r2)
forest_reg = evaluate_regression(
    "Random Forest Regressor",
    RandomForestRegressor(n_estimators=100, random_state=42),
    X_reg_train,
    X_reg_test
)
regression_table = pd.DataFrame(
    regression_results,
    columns=["Model", "MAE", "RMSE", "R2", "Training Time"]
)
print("\nRegression Model Comparison:")
display(regression_table)
coef_table = pd.DataFrame({
    "Feature": X_reg_train.columns,
    "Coefficient": linear_model.coef_
}).sort_values(by="Coefficient", ascending=False)
print("\nLinear Regression Coefficients:")
display(coef_table)
forest_reg_importance = pd.DataFrame({
    "Feature": X_reg_train.columns,
    "Importance": forest_reg.feature_importances_
}).sort_values(by="Importance", ascending=False)
print("\nRandom Forest Regressor Feature Importance:")
display(forest_reg_importance.head(10))
```

# Phase 5: Classification Modeling



```python
classification_results = []
def evaluate_classification(model_name, model, X_train_data, X_test_data):
    start_time = time.time()
    model.fit(X_train_data, y_cls_train)
    training_time = time.time() - start_time
    y_pred = model.predict(X_test_data)
    y_prob = model.predict_proba(X_test_data)[:, 1]
    accuracy = accuracy_score(y_cls_test, y_pred)
    precision = precision_score(y_cls_test, y_pred)
    recall = recall_score(y_cls_test, y_pred)
    f1 = f1_score(y_cls_test, y_pred)
    roc_auc = roc_auc_score(y_cls_test, y_prob)
    classification_results.append([model_name, accuracy, precision, recall, f1, roc_auc, training_time])
    print("\n", model_name)
    print("Accuracy:", accuracy)
    print("Precision:", precision)
    print("Recall:", recall)
    print("F1-score:", f1)
    print("ROC-AUC:", roc_auc)
    print("Training Time:", training_time)
    print("Confusion Matrix:")
    print(confusion_matrix(y_cls_test, y_pred))
    print("Classification Report:")
    print(classification_report(y_cls_test, y_pred))
    return model
log_model = evaluate_classification(
    "Logistic Regression",
    LogisticRegression(max_iter=1000, class_weight="balanced"),
    X_cls_train_scaled,
    X_cls_test_scaled
)
tree_cls = DecisionTreeClassifier(
    max_depth=5,
    min_samples_leaf=5,
    random_state=42,
    class_weight="balanced"
)
tree_cls = evaluate_classification(
    "Decision Tree Classifier",
    tree_cls,
    X_cls_train,
    X_cls_test
)
forest_cls = evaluate_classification(
    "Random Forest Classifier",
    RandomForestClassifier(n_estimators=100, random_state=42, class_weight="balanced"),
    X_cls_train,
    X_cls_test
)
classification_table = pd.DataFrame(
    classification_results,
    columns=["Model", "Accuracy", "Precision", "Recall", "F1", "ROC-AUC", "Training Time"]
)
print("\nClassification Model Comparison:")
display(classification_table)
plt.figure(figsize=(18, 8))
plot_tree(
    tree_cls,
    feature_names=X_cls_train.columns,
    class_names=["Normal", "Fraud"],
    filled=True,
    max_depth=3,
    fontsize=8
)
plt.title("Decision Tree Classifier Visualization")
plt.show()
tree_rules = export_text(tree_cls, feature_names=list(X_cls_train.columns), max_depth=3)
print("\nDecision Tree Rules:")
print(tree_rules)
forest_cls_importance = pd.DataFrame({
    "Feature": X_cls_train.columns,
    "Importance": forest_cls.feature_importances_
}).sort_values(by="Importance", ascending=False)
print("\nRandom Forest Classifier Feature Importance:")
display(forest_cls_importance.head(10))
```

# Phase 6: Model Comparison and Selection


```python
print("Regression Models:")
display(regression_table)
print("\nClassification Models:")
display(classification_table)
best_reg_model = regression_table.sort_values(by="RMSE").iloc[0]
best_cls_model = classification_table.sort_values(by="F1", ascending=False).iloc[0]
print("\nBest Regression Model:")
print(best_reg_model["Model"])
print("\nBest Classification Model:")
print(best_cls_model["Model"])
print("\nBusiness Justification:")
print("For regression, the best model is selected using lower RMSE because house price prediction should minimize large errors.")
print("For classification, the best model is selected using F1-score because fraud detection needs balance between precision and recall.")
print("Recall is especially important in fraud detection because missing a fraud transaction can cause financial loss.")
```

# Phase 7: Bias-Variance Analysis

```python
train_sizes, train_scores, test_scores = learning_curve(
    LinearRegression(),
    X_reg_train_scaled,
    y_reg_train,
    cv=3,
    scoring="r2",
    train_sizes=np.linspace(0.1, 1.0, 5)
)
train_mean = train_scores.mean(axis=1)
test_mean = test_scores.mean(axis=1)
plt.figure(figsize=(6, 4))
plt.plot(train_sizes, train_mean, marker="o", label="Training Score")
plt.plot(train_sizes, test_mean, marker="o", label="Validation Score")
plt.title("Learning Curve - Linear Regression")
plt.xlabel("Training Size")
plt.ylabel("R2 Score")
plt.legend()
plt.show()
train_sizes_cls, train_scores_cls, test_scores_cls = learning_curve(
    LogisticRegression(max_iter=1000, class_weight="balanced"),
    X_cls_train_scaled,
    y_cls_train,
    cv=3,
    scoring="f1",
    train_sizes=np.linspace(0.1, 1.0, 5)
)
train_mean_cls = train_scores_cls.mean(axis=1)
test_mean_cls = test_scores_cls.mean(axis=1)
plt.figure(figsize=(6, 4))
plt.plot(train_sizes_cls, train_mean_cls, marker="o", label="Training Score")
plt.plot(train_sizes_cls, test_mean_cls, marker="o", label="Validation Score")
plt.title("Learning Curve - Logistic Regression")
plt.xlabel("Training Size")
plt.ylabel("F1 Score")
plt.legend()
plt.show()
print("Bias-Variance Interpretation:")
print("If training score is high and validation score is low, the model may have high variance or overfitting.")
print("If both training and validation scores are low, the model may have high bias or underfitting.")
print("If both scores are close and reasonably high, the model generalizes well.")
```

# Phase 9: Hyperparameter Tuning and Optimization



```python
ridge = Ridge()
lasso = Lasso(max_iter=10000)
ridge_params = {
    "alpha": [0.01, 0.1, 1, 10, 100]
}
lasso_params = {
    "alpha": [0.01, 0.1, 1, 10, 100]
}
ridge_grid = GridSearchCV(ridge, ridge_params, cv=3, scoring="r2")
lasso_grid = GridSearchCV(lasso, lasso_params, cv=3, scoring="r2")
ridge_grid.fit(X_reg_train_scaled, y_reg_train)
lasso_grid.fit(X_reg_train_scaled, y_reg_train)
ridge_pred = ridge_grid.best_estimator_.predict(X_reg_test_scaled)
lasso_pred = lasso_grid.best_estimator_.predict(X_reg_test_scaled)
ridge_rmse = np.sqrt(mean_squared_error(y_reg_test, ridge_pred))
lasso_rmse = np.sqrt(mean_squared_error(y_reg_test, lasso_pred))
print("Ridge Best Alpha:")
print(ridge_grid.best_params_)
print("\nLasso Best Alpha:")
print(lasso_grid.best_params_)
print("\nRidge RMSE:")
print(ridge_rmse)
print("\nLasso RMSE:")
print(lasso_rmse)
rf_params = {
    "n_estimators": [50, 100],
    "max_depth": [None, 5, 10],
    "min_samples_leaf": [1, 2, 5]
}
rf_grid = GridSearchCV(
    RandomForestClassifier(random_state=42, class_weight="balanced"),
    rf_params,
    cv=3,
    scoring="f1",
    n_jobs=-1
)
rf_grid.fit(X_cls_train, y_cls_train)
best_rf_cls = rf_grid.best_estimator_
best_rf_pred = best_rf_cls.predict(X_cls_test)
best_rf_prob = best_rf_cls.predict_proba(X_cls_test)[:, 1]
print("\nBest Random Forest Classifier Parameters:")
print(rf_grid.best_params_)
print("\nTuned Random Forest Classifier Results:")
print("Accuracy:", accuracy_score(y_cls_test, best_rf_pred))
print("Precision:", precision_score(y_cls_test, best_rf_pred))
print("Recall:", recall_score(y_cls_test, best_rf_pred))
print("F1-score:", f1_score(y_cls_test, best_rf_pred))
print("ROC-AUC:", roc_auc_score(y_cls_test, best_rf_prob))
print("Confusion Matrix:")
print(confusion_matrix(y_cls_test, best_rf_pred))
print("\nFinal Optimization Summary:")
print("GridSearchCV was used to tune model hyperparameters.")
print("Ridge and Lasso were compared as regularized regression models.")
print("Random Forest Classifier was tuned to improve fraud detection performance.")
```
